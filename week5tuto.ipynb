```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer

# Create sample data
data = {
    'Age': [25, 30, np.nan, 35, 40, np.nan, 28],
    'Salary': [50000, np.nan, 70000, 80000, np.nan, 60000, 55000],
    'Experience': [2, 5, np.nan, 10, 3, 7, np.nan],
    'Department': ['IT', 'HR', 'IT', np.nan, 'Finance', 'IT', 'HR']
}

df = pd.DataFrame(data)

print("BEFORE IMPUTATION:")
print(df)
print("\nMissing values:")
print(df.isna().sum())

# --- Impute numerical columns ---
numerical_cols = ['Age', 'Salary', 'Experience']

# Impute using the mean strategy for numerical data
numerical_imputer = SimpleImputer(strategy='mean')
df[numerical_cols] = numerical_imputer.fit_transform(df[numerical_cols])

# --- Impute categorical columns ---
categorical_cols = ['Department']

# Impute using the most_frequent strategy for categorical data
categorical_imputer = SimpleImputer(strategy='most_frequent')

# Fit and transform the column
department_imputed = categorical_imputer.fit_transform(df[categorical_cols])

# Assign the imputed values back to the DataFrame
# Use .ravel() to convert the 2D array back to a 1D series for assignment
df['Department'] = department_imputed.ravel()

print("\n\nAFTER IMPUTATION:")
print(df)
print("\nMissing values:")
print(df.isna().sum())
```

```python
import pandas as pd
import numpy as np
from scipy import stats

# 1. Create a sample DataFrame with 100 rows and 3 columns of random, normally distributed data
df = pd.DataFrame(np.random.randn(100, 3))

# 2. Filter the DataFrame to keep only non-outliers
# The condition keeps rows where the absolute Z-score for ALL columns is less than 3
df_cleaned = df[((np.abs(stats.zscore(df)) < 3).all(axis=1))]

# Print the shape (rows, columns) to show how many outliers were removed
print(f"Original shape: {df.shape}")
print(f"Cleaned shape: {df_cleaned.shape}")
print("\nFirst 5 rows of the cleaned DataFrame:")
print(df_cleaned.head())
```
